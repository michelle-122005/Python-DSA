```python
# Ask user for number of days
days = int(input("Enter number of days: "))

# 1 day = 24*60*60 seconds
seconds = days * 86400

# Show the result
print(f"{days} days = {seconds} seconds")
```

```python
import math

# Ask user for the radius
radius = float(input("Enter the radius: "))

# ** means "to the power of" in Python
volume = (4/3) * math.pi * (radius ** 3)

# Show the result
print(f"The volume of the sphere is {volume:.2f}")
```

```python
def area(side):
    # Area of a square = side * side
    return side ** 2

def perimeter(side):
    # Perimeter of a square = 4 * side
    return 4 * side

# Ask user for the side length
side_length = float(input("Enter the side length of the square: "))

# Call the functions and display results
print(f"Area: {area(side_length)}")
print(f"Perimeter: {perimeter(side_length)}")
```

```python
def check_case(char):
    # isupper() and islower() are built-in checks for letters
    if char.isupper():
        return "Uppercase"
    elif char.islower():
        return "Lowercase"
    else:
        return "Not a letter"

# Ask user for a single character
user_char = input("Enter a single character: ")

# Call the function and display the result
print(check_case(user_char))
```

```python


# pseudocode step by step
x = 0
y = 20

while True:
    y -= 4          # SUBTRACT 4 FROM y
    x += 2 / y       # ADD 2/y TO x
    if y < 6:        # UNTIL y IS LESS THAN 6
        break

# Display the final value of x
print(x)
```

```python
values = []  # empty list to store the 5 numbers

# Loop 5 times, asking the user for a value each time
for i in range(5):
    num = float(input(f"Enter value {i+1}: "))
    values.append(num)

# Calculate the average
average = sum(values) / len(values)

# Display results
print(f"Values entered: {values}")
print(f"Average: {average}")
```
